```python
#如上列表中的新闻标题那些与股市相关。请用json格式输出，例如{"title":"重庆一高校多名黑人男性被曝混住在女生宿舍楼，校方：临时住一天", "relevence": False, "explain": "该新闻属于社会新闻，和股市不相关"}

# from playwright.sync_api import sync_playwright
#
# playwright = sync_playwright().start()
# browser = playwright.chromium.launch(headless=False)
# context = browser.new_context(storage_state="yuanbao_state.json")
# page = context.new_page()
# page.goto("https://yuanbao.tencent.com/", timeout=60000)
# page.wait_for_load_state('networkidle')

import asyncio
from playwright.async_api import async_playwright
import time
import json

prompt = json.dumps(articles,ensure_ascii=False) + """
请根据列表中的文章标题判断是否跟A股有关系，以json格式输出，如{'title': '【何以中国·黄河安澜】河声丨安澜之变：答案藏在那份坚定的承诺里',
  'url': 'https://m.toutiao.com/article/7550940140194071086/', "relevance": False, "explain": "社会新闻与股市无关"}, 其中title和url保持与列表中一致不要修改。"""

result = ""
async def run_playwright():
    async with (async_playwright() as p):
        browser = await p.chromium.launch(headless=False)
        context = await browser.new_context(storage_state="yuanbao_state.json")
        page = await context.new_page()
        await page.goto("https://yuanbao.tencent.com/")
        time.sleep(1)
        await page.locator("xpath=/html/body/div[2]/div/div/div[1]/div[1]/section/div[3]/div/span").click()  # 选中临时会话框
        time.sleep(1)
        await page.locator("xpath=/html/body/div[1]/div/div[2]/div[2]/div/div/div[1]/div/div[1]/div/div[3]/div/div[position()>=0 or position()<=1000]/div/div/div[2]/div[2]/div/div/div/div").fill(prompt) #fill prompt
        time.sleep(1)
        ##这里应该吧是否深度思考，是否联网设置好，否则跟随上次使用变化。
        await page.locator("xpath=/html/body/div[1]/div/div[2]/div[2]/div/div/div[1]/div/div[1]/div/div[3]/div/div[position()>=0 or position <=1000]/div/div/div[2]/div[3]/div[2]/div/a/span").click() #send
        time.sleep(60)
        btn = page.locator("xpath=/html/body/div[1]/div/div[2]/div[2]/div/div/div[1]/div/div[1]/div/div[3]/div/div[position()>=0 or position()<=1000]/div/div/div[2]/div[3]/div[2]/div/a")
        while True: #check sent button element, if button element startwith <span means llm output is over
            state = await btn.inner_html(timeout=2000)
            if state.startswith("<span"):
                break
            time.sleep(1)

        result = await page.locator("xpath=/html/body/div[1]/div/div[2]/div[2]/div/div/div[1]/div/div[1]/div/div[2]/div/div[2]/div[1]/div[3]/div/div[2]/div/div[1]/div[2]/div/div/div/div/div").text_content()

        return result
        await browser.close()

# 在 Jupyter 中直接运行
result_text = await run_playwright()
```
