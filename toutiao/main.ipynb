```python
import pandas as pd
from datetime import datetime, timedelta

df = pd.read_excel('toutiao.xlsx')
df['abs_pub_time'] = pd.to_datetime(df['abs_pub_time'])

now = pd.Timestamp.now()
seven_days_ago = now - pd.Timedelta(hours=24)

filtered_df = df[df['abs_pub_time'] > seven_days_ago]
filtered_df
```

```python
filtered_df = filtered_df.sort_values('abs_pub_time')
filtered_df
```

```python
filtered_df[['title', 'abs_pub_time']].to_json(orient='records',force_ascii=False)
```

```python
filtered_df['full_url'] = "https://www.toutiao.com"+filtered_df['url']
articles_url = filtered_df['full_url'].to_list()
len(articles_url)
```

```python
from firecrawl import Firecrawl

# firecrawl = Firecrawl(api_key="fc-YOUR-API-KEY", api_url="http://localhost:3002")
firecrawl = Firecrawl(api_key="fc-YOUR-API-KEY", api_url="http://centos.tailc67917.ts.net:3002")

start = firecrawl.start_batch_scrape(articles_url[-20:], formats=["markdown"])  # returns id

job = firecrawl.batch_scrape(articles_url[-20:], formats=["markdown"], poll_interval=2, wait_timeout=120)

print(job.status, job.completed, job.total)
```

```python
all_content = ""
for content in job.data:
    all_content = all_content+content.markdown
all_content
```

```python
import re

def remove_url_from_content(content):
    """
    从Markdown文件中删除所有图片语法
    """

    # 删除图片语法
    pattern = r'!{0,1}\[.*?\]\([^)]+\)'
    cleaned_content = re.sub(pattern, '', content)

    return cleaned_content

filtered_content = remove_url_from_content(all_content)
```

```python
import jieba  # 用于中文分词
from collections import Counter
from nltk.corpus import stopwords as nltk_stopwords
# 需先下载 nltk.download('stopwords')
stopwords = set(nltk_stopwords.words('chinese'))
custom_stopwords = ['今日头条', '头条', '今日', '打开', '查看', '评论', '图片', '详情', '阅读', '2025', '企业', '市场', '中国', '新闻', '技术', '公司','产品', '能力', '关注', '发展', 'APP','需求', '场景', '用户', '未来', '平台', '内容', '小时', '发布','资讯', '一个', '分享', '时事', '全文', '收藏', '官方', '合作','产业', '全球', '核心', '传统','领域', '城市', '行业', '提供', '价格','完整', '账号', '提升', '点击','系统','我国', '记者','很多','报名', '数字', '全国', '保障', 'https', 'www', 'toutiao', 'com', 'trending', 'rank','10','每日','晚点']
stopwords.update(custom_stopwords)

# 假设 text 是你的原始文本
words = jieba.lcut(filtered_content) # 中文分词

#过滤停用词并统计词频
filtered_words = [word for word in words if word not in stopwords and len(word) > 1] # 同时过滤掉单字
word_freq = Counter(filtered_words)

print("前10个高频词:", word_freq.most_common(50))
```

```python
wfs = []
for wf in word_freq.most_common(50):
    wfs.append(wf[0])

print(wfs)
```
